Task 3:

Rolling a dice, probability of 5 given that the rolled number is odd.

```python
total_rolls = 10000
total_odd_rolls = 0
total_5s_rolled =0

for i in range (total_rolls):
    roll = np.random.randint(1,6)
    
    if roll == 5 :
        total_5s_rolled += 1
    
    if roll % 2 !=0 :
        total_odd_rolls += 1
        
cond_prob = total_5s_rolled / total_odd_rolls

print(f"probability of rolling a 5 given that the rolled number is odd over {total_rolls} rolls = {cond_prob:.4f}")
```

output : 

"probability of rolling a 5 given that the rolled number is odd over 10000 rolls = 0.3375"

meaning that out the total odd numbered rolls, the probabilty of rolling a 5 is equal to 0.3375.
