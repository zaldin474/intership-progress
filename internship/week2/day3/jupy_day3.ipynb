Error content: 
    "ValueError: matmul: Input operand 1 has a mismatch in its core dimension 0, with gufunc signature (n?,k),(k,m?)->(n?,m?) (size 4 is different from 3)"

    The error occured due to size mismatch between z (4D vector) and x(3D vector).

    since the dot product is the length of a (projection of an array to the other array) sumed with the length of the other, when there is a size difference one vector will not have a matching vector to project on nor sum into. leaving us with an error.
 
    it can be solved by matching the sizes. In our case: (transforming the z into a 3D vector).


```python
a = np.array( [[19,10,4] , [24,9,5] ])
b = np.array( [[10,6,5] , [14,8,7]])

dot_prod_mat = np.dot(a , b)

print(dot_prod_mat)
```

Here is another form of that error: "ValueError: shapes (2,3) and (2,3) not aligned: 3 (dim 1) != 2 (dim 0)"

"Mismatched Alignment"

We are trying to calculate the dot product (matrix multiplication) of two matrices that are both 2 rows by 3 columns (2 x 3). 

When trying to multiply two matrices, the number of columns in the first matrix must match the number of rows in the second matrix.

To fix it we must trnspose the second matrix, so that its shape becomes (3 x 2) (b.T)
