```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('StudentsPerformance.csv')
```

```python
print (df.shape)
print (df.dtypes)
print (df.columns)
```

we remove some columns that are irrelevant to our work to make the data simpler.

    basically we assign a subdata to our data set (we assgin the modified on to the big one)

```python
df = df[['gender',
    #'race/ethnicity',
    'parental level of education',
    #'lunch',
       #'test preparation course'
        'math score', 'reading score',
       'writing score']]
```

Math score Plots :

```python
fig , axes = plt.subplots(1 ,2 , figsize = (18,10))

sns.histplot(df['math score'], ax = axes[0] )
sns.boxplot(df['math score'] , ax = axes[1] ) 

plt.show()
```

The histogram demonstrates a normal distribution skewed a bit right, indicating higher grades around the median.

In the above (math score) box plot we identify some outliers showen as small circles underneath the 1st Quantile. (under 27)

 we dont discard the outliers becuase they represent real value and scores of students, and discrading them would ruin the integrity of the data.

Using IQR method for (math score plot) to flag outliers:

```python
Q1 = df['math score'].quantile(0.25)
Q3 = df['math score'].quantile(0.75)
IQR = Q3 - Q1

small_whisker = Q1 - 1.5*IQR # in our case ouliers are under this value
big_whisker = Q3 + 1.5*IQR # the big whisker

outliers = df[(df["math score"] < Q1 - 1.5*IQR) | (df["math score"] > Q3 + 1.5*IQR)] #

print ("math score statistics: ")
print (f"   Small whiscker: {small_whisker}")
print (f"   Big whiscker: {big_whisker}")
print (f"   IQR: {IQR}\n")

print (outliers)
```
